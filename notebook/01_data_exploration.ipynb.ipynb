### Check: Grain of order_items and payments (Rows per Order)

```python
items_per_order = items.groupby("order_id").size()
pay_per_order = payments.groupby("order_id").size()

print("Item rows per order:")
print(items_per_order.value_counts().sort_index().head(8))

print("\nPayment rows per order:")
print(pay_per_order.value_counts().sort_index().head(8))
```

# ### Check: Fan-out Effect of a Direct Join

```python
multi_items = set(items_per_order[items_per_order > 1].index)
multi_pay = set(pay_per_order[pay_per_order > 1].index)
both = multi_items & multi_pay

print("Orders with multiple item rows    :", len(multi_items))
print("Orders with multiple payment rows :", len(multi_pay))
print("Orders with BOTH                  :", len(both))

direct_join = items.merge(payments, on="order_id")
print()
print("Rows in order_items    :", len(items))
print("Rows in payments       :", len(payments))
print("Rows after direct join :", len(direct_join))
```

```python
# **Finding:** 9,803 orders have multiple item rows, 2,961 have multiple payment rows,
# and 275 have both. A direct join of order_items (112,650 rows) and payments
# produces 117,601 rows, at least 4,951 more than the original item rows (about
# 4.4%).

# **Implication:** The extra rows are duplicates created by the join (fan-out). They
# inflate item values for orders with several payments and payment values for
# orders with several items.

# **Decision:** Aggregate order_items and payments to order level before joining,
# so each order has exactly one expected value and one payment value.
```

 ### Check: Money Impact of the Fan-out

```python
# Orders present in both tables (these survive a direct join)
common_ids = set(items["order_id"]) & set(payments["order_id"])

# Correct totals: each row counted once
correct_items = items[items["order_id"].isin(common_ids)]
correct_pay = payments[payments["order_id"].isin(common_ids)]

correct_item_total = (correct_items["price"] + correct_items["freight_value"]).sum()
correct_pay_total = correct_pay["payment_value"].sum()

# Totals after a direct join (rows get multiplied)
direct_join = items.merge(payments, on="order_id")
join_item_total = (direct_join["price"] + direct_join["freight_value"]).sum()
join_pay_total = direct_join["payment_value"].sum()

print(f"Item total    | correct: {correct_item_total:>14,.2f} | after join: {join_item_total:>14,.2f}")
print(f"Payment total | correct: {correct_pay_total:>14,.2f} | after join: {join_pay_total:>14,.2f}")
print()
print(f"Item total inflated by    : {join_item_total - correct_item_total:,.2f}")
print(f"Payment total inflated by : {join_pay_total - correct_pay_total:,.2f}")
```

```python
# *Finding:** A direct join inflates the item total by R$ 723,134 (about 4.6%) and the
# payment total by R$ 4,461,855 (about 28.2%). At the correct grain, item and
# payment totals are almost equal (R$ 15,843,410 vs R$ 15,846,280, a gap of
# R$ 2,870). After the direct join, a false gap of about R$ 3.74 million appears.

# **Implication:** Joining without aggregation creates fake reconciliation
# differences. Matching totals do not prove that every order matches, since
# over- and underpayments can offset each other.

# **Decision:** Aggregate order_items and payments to order level before joining,
# then compare order by order
```

### Check: The 3 not_defined Orders and the Largest Order

```python
check_ids = list(payments.loc[payments["payment_type"] == "not_defined", "order_id"])
check_ids.append(payments.sort_values("payment_value", ascending=False).iloc[0]["order_id"])

for oid in check_ids:
    status = orders.loc[orders["order_id"] == oid, "order_status"].values[0]
    item_rows = items[items["order_id"] == oid]
    item_total = (item_rows["price"] + item_rows["freight_value"]).sum()
    pay_total = payments.loc[payments["order_id"] == oid, "payment_value"].sum()
    print(f"{oid[:8]} | status: {status:10} | item rows: {len(item_rows)} | "
          f"item total: {item_total:10,.2f} | payment total: {pay_total:10,.2f}")
```

```python

# **Finding:** The 3 `not_defined` orders are all `canceled`, with no item rows and
# zero payment, so there is nothing to reconcile. The order with the largest
# payment (R$ 13,664.08) is `delivered`, has 8 item rows, and its item total
# matches the payment exactly.

# **Implication:** The `not_defined` rows are not real underpayments, and the large
# payment is a legitimate large order. This order also shows the fan-out risk:
# a direct join would repeat its single payment 8 times (about R$ 109,313).

# **Decision:** No cleaning needed for these rows. Keep them, and treat canceled
# orders with no items and no payment as expected rather than as exceptions.
```

### Check: Order Status and Financial Behavior

```python
# Order-level totals (aggregate first, then join, to avoid fan-out)
expected = (
    items.assign(item_value=items["price"] + items["freight_value"])
    .groupby("order_id")["item_value"].sum()
    .rename("expected_value")
)
paid = payments.groupby("order_id")["payment_value"].sum().rename("paid_value")

summary = (
    orders[["order_id", "order_status"]]
    .merge(expected, on="order_id", how="left")
    .merge(paid, on="order_id", how="left")
    .fillna({"expected_value": 0, "paid_value": 0})
)

by_status = summary.groupby("order_status").agg(
    orders=("order_id", "count"),
    expected_total=("expected_value", "sum"),
    paid_total=("paid_value", "sum"),
).round(2)

by_status["difference"] = (by_status["paid_total"] - by_status["expected_total"]).round(2)
by_status
```

```python
# **Finding:** Delivered orders (96,478, about 97% of all orders) reconcile almost
# exactly: payments exceed item value by R$ 2,688 (about 0.017%). In contrast,
# `unavailable` orders show payments of R$ 126,480 against item value of only
# R$ 2,140 (+R$ 124,339), and `canceled` orders show +R$ 37,370. Together these two
# statuses are about 1.2% of orders but account for about 97.8% of the total net
# difference (R$ 165,319). Payments are never below item value in any status.

# **Implication:** Reconciliation differences are concentrated in canceled and
# unavailable orders, not in delivered ones. The data has no negative payments,
# so any refunds are not visible and these amounts cannot be called leakage.

# **Decision:** Analyze delivered orders separately from canceled and unavailable
# orders, and report the latter as exposure to investigate.
```
