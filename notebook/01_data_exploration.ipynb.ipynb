# Check: Order Coverage Across Tables

```python
order_ids = set(orders["order_id"])
item_ids = set(items["order_id"])
pay_ids = set(payments["order_id"])

print("Unique orders in orders      :", len(order_ids))
print("Unique orders in order_items :", len(item_ids))
print("Unique orders in payments    :", len(pay_ids))
print()
print("Orders with NO items         :", len(order_ids - item_ids))
print("Orders with NO payments      :", len(order_ids - pay_ids))
print("Item orders not in orders    :", len(item_ids - order_ids))
print("Payment orders not in orders :", len(pay_ids - order_ids))
```

```python
# *Finding:** Of 99,441 orders, 775 have no items and 1 has no payment record.
# There are no orphan records: every item and payment belongs to an existing order.

# **Implication:** These orders are the first reconciliation exceptions (`No items`,
# `No payment`). A left join starting from `orders` is needed so they are not
# silently dropped.
```

```python
# ### Check: Order Status of Orders With No Items
no_items_ids = order_ids - item_ids
orders_no_items = orders[orders["order_id"].isin(no_items_ids)]

print(orders_no_items["order_status"].value_counts())
```

```python
# **Finding:** Of the 775 orders with no items, 767 (about 99%) are `unavailable` (603)
# or `canceled` (164). The remaining 8 are `created` (5), `invoiced` (2) and
# `shipped` (1).

# **Implication:** For cancelled and unavailable orders, missing items are expected
# business behavior, not data errors. The 8 orders in other statuses are unusual
# and are flagged in the data quality log.

# **Decision:** Keep all 775 orders. They will fall in the `No items` class, and
# cancelled and unavailable orders will be identified separately in the analysis.
```

### Check: Grain of order_items and payments (Rows per Order)

```python
items_per_order = items.groupby("order_id").size()
pay_per_order = payments.groupby("order_id").size()

print("Item rows per order:")
print(items_per_order.value_counts().sort_index().head(8))

print("\nPayment rows per order:")
print(pay_per_order.value_counts().sort_index().head(8))
```

# ### Check: Fan-out Effect of a Direct Join

```python
multi_items = set(items_per_order[items_per_order > 1].index)
multi_pay = set(pay_per_order[pay_per_order > 1].index)
both = multi_items & multi_pay

print("Orders with multiple item rows    :", len(multi_items))
print("Orders with multiple payment rows :", len(multi_pay))
print("Orders with BOTH                  :", len(both))

direct_join = items.merge(payments, on="order_id")
print()
print("Rows in order_items    :", len(items))
print("Rows in payments       :", len(payments))
print("Rows after direct join :", len(direct_join))
```

```python
# **Finding:** 9,803 orders have multiple item rows, 2,961 have multiple payment rows,
# and 275 have both. A direct join of order_items (112,650 rows) and payments
# produces 117,601 rows, at least 4,951 more than the original item rows (about
# 4.4%).

# **Implication:** The extra rows are duplicates created by the join (fan-out). They
# inflate item values for orders with several payments and payment values for
# orders with several items.

# **Decision:** Aggregate order_items and payments to order level before joining,
# so each order has exactly one expected value and one payment value.
```
